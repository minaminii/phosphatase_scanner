In the context of sequence similarity searches (like HMMER or BLAST), an E-value (expectation value) is a statistical measure that tells you:

> 👉 How many hits you would expect to see by chance with a score at least as good as the one observed, given the size of the database.

You buy 1 lottery ticket. - The chance of winning is 1 in a million → that’s like a p-value of 1e-6.

Now suppose 10 million people each buy a ticket.

- On average, you’d expect about 10 winners by pure chance.
- That “expected number of winners” is the E-value.

Low E-value → strong evidence that the match is real (not random).

High E-value → more likely the match is just noise.


Rules of thumb:

E < 1e-3 → usually considered significant.

E < 1e-6 → very strong match.

E ~ 1 or higher → likely random.

So, when your function filters on min_evalue=1e-6, it’s keeping only the very confident matches.

```python

def classify_hmmscan(hmmscan_tbldf: pd.DataFrame, min_evalue=1e-6):
    """
    Classify HMMER hmmscan hits into confidence categories.

    This function takes the tabular output from `read_hmmscan_tblout`, filters out
    hits with very weak E-values, and assigns each hit to one of the following
    categories based on heuristic thresholds:

        - "High-confidence"
        - "Probable"
        - "Weak"
        - "Fragmentary"
        - "Ambiguous"
        - "No match"

    Parameters
    ----------
    hmmscan_tbldf : pandas.DataFrame
        DataFrame containing hmmscan results, as formatted by
        `read_hmmscan_tblout`. Must include the following columns:
        ``["full_seq_Evalue", "best_dom_Evalue", "best_dom_score", "best_dom_bias",
        "exp", "clu", "ov", "env", "rep", "inc"]``.
    min_evalue : float, optional
        Minimum full-sequence E-value threshold for filtering hits (default = 1e-6).

    Returns
    -------
    pandas.DataFrame
        A copy of the input DataFrame with an additional column ``"confidence"``
        containing the assigned category for each hit.

    Raises
    ------
    BaseException
        If one or more of the required columns are missing from the input DataFrame.

    Notes
    -----
    The classification scheme is heuristic and may need to be adjusted for
    specific datasets or applications.
    """
    _df = hmmscan_tbldf.copy()
    # Filter out very weak hits (full_seq_Evalue < 1e-6)
    _df = _df.loc[_df['full_seq_Evalue'] < min_evalue]
    expected_cols = [
        "full_seq_Evalue",
        "best_dom_Evalue", "best_dom_score", "best_dom_bias",
        "exp", "clu", "ov", "env", "rep", "inc",
    ]
    # Check if all of the required columns exists in the input dataframe.
    if set(expected_cols).intersection(set(_df.columns)) != set(expected_cols):
        raise BaseException('classify_hmmscan: Expected column(s) not found in dataframe.')

    def classify(row):
        full_e = row["full_seq_Evalue"]
        dom_e = row["best_dom_Evalue"]
        score = row["best_dom_score"]
        bias = row["best_dom_bias"]
        inc = row["inc"]
        rep = row["rep"]
        exp = row["exp"]
        env = row["env"]
        clu = row["clu"]
        ov = row["ov"]

        # Decision tree
        if full_e <= 1e-5:
            if inc >= 1:
                if dom_e <= 1e-3 and bias < 0.1 * score:
                    return "High-confidence"
                else:
                    return "Probable"
            else:
                if dom_e <= 1e-3 and rep >= 1:
                    return "Probable"
                elif exp >= 0.5 and env > 0:
                    return "Weak"
                else:
                    return "Fragmentary"
        else:
            if dom_e <= 1e-3 and inc >= 1:
                return "Probable"
            elif clu > 0 or ov > 0:
                return "Ambiguous"
            elif rep >= 1 and dom_e <= 1e-2:
                return "Weak"
            else:
                return "No match"

    _df["confidence"] = _df.apply(classify, axis=1)
    return _df
```

1) Is the full-sequence E-value ≤ 1e-5?

Yes → (2)

No → (7)

2) Does the hit have at least one domain with inc ≥ 1 (domain included)?

Yes → (3)

No → (4)

3) Is the best domain independent E-value (i-Evalue) ≤ 1e-3 and bias << score (bias not large)?

Yes → High-confidence match (stop)

No → Probable match (stop)

4) (full E ≤1e-5 but no included domains) Is the best domain conditional E-value (c-Evalue) ≤ 1e-3 AND rep ≥ 1?

Yes → Probable match (stop)

No → (5)

5) Does exp (expected domains) ≥ 0.5 and env/dom > 0 (envelope exists) and alignment covers ≥ 50% of canonical domain length?

Yes → Weak match (stop)

No → Fragmentary / Partial match (stop)

6) — (not used; placeholder)

7) (full-sequence E-value > 1e-5) Is best domain i-Evalue ≤ 1e-3 AND inc ≥ 1?

Yes → Probable match (stop)

No → (8)

8) Are there overlapping/clustered domains (clu > 0 or ov > 0) or env shows multiple conflicting envelopes?

Yes → Ambiguous / overlapping — manual review required (stop)

No → (9)

9) Is any domain rep ≥ 1 with c-Evalue ≤ 1e-2 and alignment includes key conserved motif(s) (e.g., catalytic residues)?

Yes → Weak match — validate motifs (stop)

No → No match (stop)

Quick numeric thresholds (adjust if you want stricter/looser)

- Full-sequence: strong ≤ 1e-5, moderate ≤ 1e-3, otherwise weak.
- Domain i-Evalue (independent): strong ≤ 1e-3.
- Domain c-Evalue (conditional): use to rescue borderline cases (≤ 1e-2).
- Inclusion: inc ≥ 1 = trustworthy domain call.
- Alignment coverage: ≥ 50% of model length useful; ≥ 80% desirable.
- Bias: if bias > ~10% of score or absolute bias large, treat with caution.

Short recommended actions per class

- High-confidence — annotate automatedly; include domain annotation and conserved-site notes.
- Probable — accept, but flag for motif/reciprocal check.
- Weak — require manual inspection of alignment and motifs; consider reciprocal BLAST/Pfam search.
- Fragmentary/Partial — may be truncated protein or poor model match — try longer sequence, check contig ends, or relax model.
- Ambiguous/Overlapping — manually inspect envelope coordinates, try alternative models, or re-run with different thresholds.
- No match — ignore for that family; consider other families or de novo analysis.

```python
def plot_fold_family_proportion(df, column="fold_family", title="Fold family distribution"):
    """
    Plot a circular (pie) chart showing the proportion of fold families.

    Parameters
    ----------
    df : pandas.DataFrame
        DataFrame containing fold family annotations.
    column : str
        Column name holding fold family labels.
    title : str
        Title of the plot.
    """
    counts = df[column].value_counts()

    plt.figure()
    plt.pie(
        counts.values,
        labels=counts.index,
        autopct='%1.1f%%',
        startangle=90,
    )
    plt.title(title)
    plt.axis('equal')  # ensures circle, not ellipse
    plt.tight_layout()
    plt.show()
```

```python
import re
from IPython.display import display, HTML

def highlight_motifs(seq, pattern=r"HC.....R", color="yellow"):
    """
    Highlight all regex matches in a sequence with <span>.
    
    Args:
        seq (str): protein sequence
        pattern (str): regex pattern for motif (default = HCX5R)
        color (str): highlight background color
    
    Returns:
        HTML object with highlighted sequence
    """
    highlighted = ""
    last_end = 0
    
    for m in re.finditer(pattern, seq):
        start, end = m.start(), m.end()
        # add unmodified part
        highlighted += seq[last_end:start]
        # add highlighted part
        highlighted += f'<span style="background-color:{color};">{seq[start:end]}</span>'
        last_end = end
    # add the rest
    highlighted += seq[last_end:]
    
    return HTML(f"<pre style='font-family:monospace;'>{highlighted}</pre>")

# Example usage
# sequence = "MTLTAHCXXXXXRVIPLCASWFHCYYYYYRGS"
sequence = str(matched_sequences[5].seq)
display(highlight_motifs(sequence, pattern=r"H?C.....R", color="yellow"))
```

```python
def expand_motif(length, constrained):
    """
    constrained: dict[pos] -> set of allowed residues
    """
    motif = [None] * length
    for pos, allowed in constrained.items():
        motif[pos] = set(allowed)
    return motif

CC1_MOTIFS = [
    expand_motif(8, {0: {'H'}, 1: {'C'}, 7: {'R'}}),
    expand_motif(8, {1: {'C'}, 7: {'R'}}),
]
CC3_MOTIFS = CC1_MOTIFS

def motif_mismatches(seq, motif):
    """
    Returns list of (position, expected, observed)
    """
    mismatches = []
    for i, allowed in enumerate(motif):
        if allowed is None:
            continue
        if seq[i] not in allowed:
            mismatches.append((i, allowed, seq[i]))
    return mismatches

def find_near_matches(sequence, motif, max_mismatches=1):
    '''
    Sliding scan to find best near-match
    '''
    hits = []
    L = len(motif)
    for i in range(len(sequence) - L + 1):
        window = sequence[i:i+L]
        mismatches = motif_mismatches(window, motif)
        if len(mismatches) <= max_mismatches:
            hits.append({
                "start": i,
                "seq": window,
                "mismatches": mismatches,
            })
    return hits

def connect_hits(A_hits, B_hits, max_gap=200):
    '''
    Connected motifs (PPM, HP): split + reconnect
    '''
    connected = []
    for a in A_hits:
        a_end = a["start"] + len(a["seq"])
        for b in B_hits:
            if b["start"] > a_end and (b["start"] - a_end) <= max_gap:
                connected.append({
                    "A": a,
                    "B": b,
                    "gap": b["start"] - a_end,
                })
    return connected

sequence = str(matched_sequences[0].seq)
find_near_matches(sequence, CC3_MOTIFS, 1)
```

```python
def levenshteinFullMatrix(str1, str2):
    m = len(str1)
    n = len(str2)

    # Initialize a matrix to store the edit distances
    dp = [[0 for _ in range(n + 1)] for _ in range(m + 1)]

    # Initialize the first row and column with values from 0 to m and 0 to n respectively
    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j

    # Fill the matrix using dynamic programming to compute edit distances
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if str1[i - 1] == str2[j - 1]:
                # Characters match, no operation needed
                dp[i][j] = dp[i - 1][j - 1]
            else:
                # Characters don't match, choose minimum cost among insertion, deletion, or substitution
                dp[i][j] = 1 + min(dp[i][j - 1], dp[i - 1][j], dp[i - 1][j - 1])

    # Return the edit distance between the strings
    return dp[m][n]

# Driver code
str1 = "HM.....R"
str2 = "HC.....R"

# Function Call
distance = levenshteinFullMatrix(str1, str2)
print(f"Levenshtein Distance: {distance}")
```

```python
def mask_to_pattern(s: str, mask: str, placeholder: str = ".") -> str:
    """Apply a mask to string `s` where:
       - 'x' means keep the character
       - '?' means replace with placeholder
    """
    return "".join(
        ch if m == "x" else placeholder
        for ch, m in zip(s, mask)
    )
print(mask_to_pattern('HMLPTKER', 'xx?????x'))
```

```python
def chop_to_window(query, window_size, step):
    """
    Generate sliding windows from `query`.

    Args:
        query (str | list): the input sequence
        window (int): length of each window
        step (int): step size to move the window

    Yields:
        str | list: window slices
    """
    for i in range(0, len(query) - window_size + 1, step):
        yield query[i:i + window_size]

def windowed_scan(query, window_size, step, mask, ref_pattern):
    results = []
    for window in chop_to_window(query, window_size, step):
        masked = mask_to_pattern(window, mask)
        distance = levenshteinFullMatrix(masked, ref_pattern)
        results.append({
            'query': window,
            'distance': distance
        })
    return results
```
